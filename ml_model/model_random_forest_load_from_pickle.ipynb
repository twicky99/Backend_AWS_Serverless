```python
import pandas as pd
import numpy as np
import pickle
import bz2
import _pickle as cPickle
```

```python
# Load the compressed Model back from file
def decompress_pickle(file):
    data = bz2.BZ2File(file, 'rb')
    data = cPickle.load(data)
    return data
```

```python
Pickled_RF_Model = decompress_pickle('compressed.pbz2')
```

```python
# Predict using the loaded model
# 'car','Toyota','Allion 240',2004,'Automatic','Petrol',1500,140000,4150000
# New data to predict the vehicle value
new_data = ['car','Toyota','Vitz',2016,'Automatic','Petrol',1000,63000,4425000]

# Creating a list with the columns in the original dataframe
new_columns = ['category','make','model','model_year','gear','fuel_type','engine_cc','mileage_km','price']

# Create a dataframe for the single vehicle
single_df = pd.DataFrame(np.array(new_data).reshape(1,len(new_data)), columns=new_columns)

# Predict the value for the single vehicle
single_predict = Pickled_RF_Model.predict(single_df.drop("price",axis=1))
single_predict[0]
```

```python
# Error rate of a single prediction
actual = float(np.array(single_df['price'])[0])
pred = single_predict[0]
diff = actual-pred
if diff < 0:
    diff = diff * -1
perc = diff/actual*100
# print(single_df.iloc[0])
print('Actual is          : ' + str(actual))
print('Prediction is      : ' + str(round(pred,2)))
print('Difference is      : ' + str(round(diff,2)))
print('Error Percentage is: ' + str(round(perc,2))+'%')
```
